# Fixer - Complete Gameplay Design Specification

I'm building a multiplayer cyberpunk roguelike browser game called "Fixer". This document contains the complete gameplay design. Build the technical implementation however you think is best.

## Core Concept

Players are mercenary "fixers" in a persistent, procedurally generated cyberpunk megacity. They spawn with nothing, complete dangerous contracts to earn credits and gear, engage in PvP through assassination mechanics, and can purchase housing to permanently save their character. Death means losing everything unless you've secured housing.

---

## World Design

### The Infinite City

The city generates infinitely as players explore it. The world is persistent - once a chunk is generated, it stays the same for all players forever. New players spawn at random locations within the already-generated city.

**District Types:**
- **Corporate Zones**: Gleaming skyscrapers, high security, wealthy targets, premium gear drops
- **Industrial Sectors**: Factories, warehouses, heavy machinery, bulk resources
- **Residential Areas**: Apartment blocks, mixed security, civilian population, housing available
- **Underground Markets**: Black markets, illegal tech, low security, dangerous NPCs
- **Wasteland**: Ruined areas, abandoned buildings, scavenger gangs, rare loot

Each district has:
- Unique building types and densities
- Different NPC factions and enemy types
- Varying security levels and police response
- District-specific loot tables
- Environmental hazards and atmosphere

### Building Interiors

Players can enter buildings. Each building has:
- Multiple floors (1-30+ depending on building type)
- Procedurally generated room layouts
- Security systems (cameras, locks, guards)
- Loot containers and hackable terminals
- NPCs (civilians, guards, other fixers)
- Environmental storytelling elements

## Player Progression

### The Death Loop

Every new player spawns as a fresh character with:
- Basic clothing
- 100 health
- 0 credits
- Starting pistol with limited ammo
- No augmentations

**The Goal**: Survive long enough to earn housing

**Housing Costs:**
- Slum Apartment: 5,000 credits
- Residential Unit: 15,000 credits
- Corporate Suite: 50,000 credits

Once you purchase housing, your character persists between sessions. Without housing, death means complete restart.

### Character Persistence

**With Housing:**
- Character saves when you log out from your apartment
- Inventory persists between sessions
- Augmentations remain installed
- Credits are stored in your apartment safe
- You can choose to venture out or stay safe

**Death With Housing:**
- You respawn at your apartment
- Keep everything stored at home
- Lose whatever you were carrying when you died
- Other players can loot your corpse

**Death Without Housing:**
- Complete character reset
- Back to main menu
- Start from scratch with new spawn location

## Contract System

### AI Fixers

The city is divided into territories controlled by procedurally generated AI "fixers" - crime bosses, corporate handlers, gang leaders. Each has:
- Unique name and personality
- Territory they control
- Preferred contract types
- Payment rates
- Reputation levels with player

Players receive contracts via cyber-augment communication (always in-ear, can't disable).

### Contract Types

**1. Assassination Contracts (Player vs Player)**
- You receive target's name and last known district
- 30-minute time limit to complete the kill
- Target is notified after 2 minutes (grace period)
- High credit rewards (5,000-20,000 based on target's gear value)
- Target drops all carried items on death
- If timer expires, contract fails, you lose reputation

**Target Experience:**
- Notification: "CONTRACT ACTIVE - YOU ARE BEING HUNTED"
- No exact location of hunter shown
- Can see hunter if they're in line of sight
- Can use stealth augments to hide
- Can complete other contracts while being hunted
- Surviving 30 minutes gives bonus credits and reputation

**2. Data Extraction**
- Infiltrate specific building
- Hack terminal to download data
- Avoid/eliminate security
- Extract without dying
- Time-sensitive (alarm systems)
- Rewards: 1,000-8,000 credits + rare gear chance

**3. Territory Control**
- Clear gang stronghold of all enemies
- Hold position for 10 minutes
- Defend against waves
- Other players may interfere
- Rewards: 3,000-12,000 credits + territory unlocks

**4. Sabotage Operations**
- Destroy corporate infrastructure
- Plant explosives or viruses
- Escape before security arrives
- Affects entire district temporarily (blackouts, chaos)
- Rewards: 2,000-10,000 credits

**5. Escort Missions**
- Protect VIP NPC through dangerous territory
- Other players may have contracts to kill your VIP
- VIP follows at walking speed, no combat ability
- Must reach destination alive
- Rewards: 4,000-15,000 credits

**6. Recovery Operations**
- Locate crashed vehicle/stolen goods
- Return items to drop-off point
- Other players may be hunting same objective
- First to deliver gets full reward
- Rewards: 2,000-9,000 credits + recovered items

**7. Surveillance Contracts**
- Observe target location without being detected
- Take photos of specific targets/events
- Remain hidden for duration
- Can't engage in combat without failing
- Rewards: 1,500-6,000 credits

**8. Assassination Contracts (NPC)**
- Eliminate high-value NPC targets
- Targets have bodyguards and security
- Stealth or loud approaches viable
- Police response on high-profile kills
- Rewards: 3,000-15,000 credits + target's gear

## Combat System

### Real-Time Combat

**Core Mechanics:**
- Top-down shooter with mouse aim
- WASD movement
- Left click to shoot
- Right click for weapon alt-fire/aim
- Space bar for dodge roll (stamina cost)

**Line of Sight:**
- Vision cones for NPCs and security systems
- Fog of war - only see what's in your field of view
- Muzzle flashes reveal position
- Sound propagates and alerts nearby enemies

**Cover System:**
- Duck behind objects for protection
- Blind fire over/around cover (less accurate)
- Cover can be destroyed
- Movement between cover is crucial

**Weapon Types:**
- Pistols: Low damage, high accuracy, quiet
- SMGs: High fire rate, low accuracy, medium damage
- Rifles: High damage, medium fire rate, loud
- Shotguns: Devastating close range, terrible distance
- Sniper rifles: One-shot potential, slow fire rate
- Energy weapons: Unique effects, ammo is battery charge
- Cyber-weapons: Installed augments that deploy weapons

**Damage System:**
- Health: 100 base (augmentable to 200+)
- Armor: Absorbs percentage of damage
- Headshots: 2x damage multiplier
- Bleeding: Damage over time from heavy wounds
- Limb damage: Affects movement/aim

**Death:**
- 5-second bleed-out period
- Can be revived by teammate (if in group)
- Drop all inventory items on death
- Corpse remains for 10 minutes
- Other players can loot corpse

## Stealth System

### Core Stealth Mechanics

**Detection States:**
- Unaware: NPCs/players don't see you
- Suspicious: Something caught attention, investigating
- Alert: Fully aware, combat engaged
- Searching: Lost sight, actively hunting

**Stealth Actions:**
- Crouch: 50% speed, reduced noise, harder to spot
- Takedown: Silent melee kill from behind (1-second animation)
- Drag bodies: Hide evidence from patrols
- Lockpicking: Open doors silently (time-based mini-game)
- Vent crawling: Access restricted areas

**Detection Factors:**
- Light level (shadows provide concealment)
- Noise level (running makes sound, crouching quiet)
- Line of sight (obstacles block vision)
- NPC alertness level (higher security = more aware)

**Stealth Augmentations:**
- **Thermoptic Camo**: 30-second invisibility (cooldown: 2 minutes)
- **Sound Dampeners**: Silent movement even when running
- **Radar Cloak**: Invisible to electronic detection
- **Wall-Hack Vision**: See enemies through walls
- **Holographic Decoy**: Create distraction clone

## Hacking System

### Hackable Targets

**Security Cameras:**
- Disable: Blind security temporarily
- Loop: Show old footage, sneak past
- Control: View camera feeds remotely
- Hack time: 3-5 seconds

**Door Locks:**
- Electronic locks: Hack to unlock (5-10 seconds)
- Biometric locks: Need keycard or biological sample
- Reinforced locks: Require advanced cyberdeck
- Hack time: 5-15 seconds based on security

**Terminals:**
- Financial data: Steal credits
- Security systems: Disable alarms/turrets
- Intel databases: Reveal contracts/locations
- Employee records: Get building layouts
- Hack time: 10-30 seconds based on ICE

**Turrets & Drones:**
- Disable: Turn off temporarily
- Reprogram: Fight for you temporarily
- Overload: Explode and damage nearby enemies
- Hack time: 8-15 seconds

**Enemy Augmentations:**
- Scramble: Disable target's cyberware
- Hijack: Take control of target's movement (3 seconds)
- Overheat: Damage target through their implants
- Hack time: 5-8 seconds (combat hacking)

### Hacking Gameplay

**Cyberdeck Requirement:**
- Must have cyberdeck equipped
- Better decks = faster hacks, higher success rate
- Deck can be upgraded with programs

**ICE (Security Programs):**
- Levels 1-5 difficulty
- Failed hacks alert security
- Some systems have counter-hacking that damages you
- Trace programs try to locate your position

**Hacking Mini-Game:**
- Circuit completion puzzle (simple but tense)
- Time pressure increases with security level
- Can be interrupted by combat/detection
- Multiple attempts possible but increase alert level

**Remote Hacking:**
- Hack cameras/systems from distance
- Must be within range (50 meters)
- Can chain hacks through networked systems
- High-level play: hack entire building from one terminal
